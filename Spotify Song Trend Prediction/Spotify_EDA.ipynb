🎵 Track Information

1. track_id → A unique identifier assigned to each song.
2. artists → The performer(s) of the track (can include multiple artists).
3. album_name → The name of the album that the song belongs to.
4. track_name → The title of the song.
5. track_genre → The genre or category of the track (e.g., pop, rock, jazz).

🌟 Popularity & Duration

1. popularity → A score (0–100) showing how popular a track is on Spotify, based on streams and user activity.
2. duration_ms → The total length of the song in milliseconds.

⚠️ Content

1. explicit → Indicates whether the track has explicit (adult or offensive) content — True or False.

🎚️ Audio Features

1. These are quantitative values derived from Spotify’s audio analysis API, describing the sound and feel of a song:
2. danceability → How suitable a track is for dancing (0.0 to 1.0).
3. energy → Intensity or activity level of a song (0.0 to 1.0).
4. key → The musical key the track is in (0–11, representing C to B).
5. loudness → Average loudness of the track in decibels (dB).
6. mode → Indicates whether the song is in a major (1) or minor (0) key.
7. speechiness → Detects the presence of spoken words (higher = more speech-like).
8. acousticness → Confidence measure of whether a track is acoustic (0.0 to 1.0).
9. instrumentalness → Likelihood that a track has no vocals (closer to 1 = more instrumental).
10. liveness → Probability that the track was recorded live (higher = more audience presence).
11. valence → Describes the musical “positiveness” of a track (0.0 = sad, 1.0 = happy).
12. tempo → Estimated speed of the song in beats per minute (BPM).
13. time_signature → Estimated beats per bar (common values: 3, 4, or 5).



This is the description of each columns along with their general values. We will see whether the dataset is diverted from this info or not.

## feature engineering

```python
df['popularity_level'] = pd.cut(df['popularity'],
                                bins=[0, 40, 70, 100],
                                labels=['Low', 'Medium', 'High'])

df['popularity_level'].value_counts()
```

```python
df['duration_category'] = pd.cut(df['duration_min'],
                                 bins=[0, 2.5, 4.5, 15],
                                 labels=['Short', 'Medium', 'Long'],
                                 include_lowest=True)
df['duration_category'].value_counts()
```

```python
df['mood_intensity'] = (df['valence'] * df['energy']) - (df['loudness'] / 60)
df['mood_intensity']
```

```python
df['replay_value_score'] = (
    df['danceability']*0.35 + 
    df['energy']*0.35 + 
    (1 - df['instrumentalness'])*0.20 +
    (df['duration_min'].between(2.5, 4.5).astype(int) * 0.10)
)
df['replay_value_score']
```

```python
df['party_index'] = df['energy']*0.4 + df['danceability']*0.4 + (1 - df['speechiness'])*0.2
df['party_index']
```

```python
df['sadness_score'] = (1 - df['valence']) * (1 - df['energy'])
df['sadness_score']
```

```python
df['production_complexity'] = (
    df['instrumentalness']*0.5 + 
    df['speechiness']*0.3 + 
    df['liveness']*0.2
)
df['production_complexity']
```

```python
df['artist_popularity_deviation'] = df['popularity'] - df.groupby('artists')['popularity'].transform('mean')
df['artist_popularity_deviation']
```

```python
df['vibiness_index'] = (
    df['danceability']*0.4 + 
    df['valence']*0.3 + 
    (df['tempo'].between(90,140).astype(int) * 0.3)
)
df['vibiness_index']
```

```python
df['skip_likelihood'] = (
    (df['duration_min'] > 5).astype(int)*0.3 +
    (1 - df['energy'])*0.3 + 
    (1 - df['danceability'])*0.2 +
    df['instrumentalness']*0.2
)
df['skip_likelihood']
```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Step 1: Choose numeric columns for PCA (only continuous ones)
pca_features = [
    'duration_min', 'loudness', 'danceability', 'energy',
    'speechiness', 'acousticness', 'instrumentalness',
    'liveness', 'valence', 'tempo'
]

# Step 2: Standardize data (PCA is variance-based)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df[pca_features])

# Step 3: Apply PCA (retain enough components for 90% variance)
pca = PCA(n_components=0.9)
pca_data = pca.fit_transform(scaled_data)

# Step 4: Create PCA DataFrame
pca_df = pd.DataFrame(
    pca_data,
    columns=[f'PC{i+1}' for i in range(pca_data.shape[1])]
)

# Step 5: Combine with target (e.g., popularity or genre)
if 'track_genre' in df.columns:
    pca_df['track_genre'] = df['track_genre']
if 'popularity' in df.columns:
    pca_df['popularity'] = df['popularity']

# Step 6: Explained variance plot
plt.figure(figsize=(8,6))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA - Explained Variance by Components')
plt.grid(True)
plt.show()

# Step 7: Visualize first 2 components (colored by popularity)
plt.figure(figsize=(8,6))
sns.scatterplot(
    x=pca_df['PC1'], y=pca_df['PC2'],
    hue=pca_df['popularity'], palette='viridis', alpha=0.6
)
plt.title('PCA: Songs Projected in 2D (colored by Popularity)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Popularity', loc='best')
plt.show()

# Step 8: Optional - view component loadings
loadings = pd.DataFrame(
    pca.components_.T,
    columns=[f'PC{i+1}' for i in range(pca.n_components_)],
    index=pca_features
)
print("\nPCA Component Loadings:\n")
print(loadings.round(3))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
from sklearn.cluster import KMeans
from imblearn.over_sampling import SMOTE
import joblib

df = pd.read_csv("spotify_cleaned.csv")
threshold = df['popularity'].quantile(0.6)
df['hit'] = (df['popularity'] >= threshold).astype(int)

features = [
    'duration_ms', 'explicit', 'danceability', 'energy', 'key', 'loudness',
    'mode', 'speechiness', 'acousticness', 'instrumentalness',
    'liveness', 'valence', 'tempo', 'time_signature'
]

df['duration_min'] = df['duration_ms'] / 60000
df['energy_valence_ratio'] = df['energy'] / (df['valence'] + 0.01)
df['acousticness_inverse'] = 1 - df['acousticness']
df['dance_energy_mix'] = (df['danceability'] + df['energy']) / 2
df['tempo_norm'] = (df['tempo'] - df['tempo'].mean()) / df['tempo'].std()

X = df[features + [
    'duration_min', 'energy_valence_ratio', 'acousticness_inverse',
    'dance_energy_mix', 'tempo_norm'
]]
y = df['hit']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=0.9, random_state=42)
X_pca = pca.fit_transform(X_scaled)

X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42)
X_res, y_res = SMOTE(random_state=42).fit_resample(X_train, y_train)

model = RandomForestClassifier(
    n_estimators=500,
    max_depth=18,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight='balanced_subsample',
    n_jobs=-1,
    random_state=42
).fit(X_res, y_res)

y_prob = model.predict_proba(X_test)[:, 1]
y_pred = (y_prob >= 0.35).astype(int)

print("Accuracy:", round(accuracy_score(y_test, y_pred), 3))
print("F1:", round(f1_score(y_test, y_pred), 3))
print("ROC AUC:", round(roc_auc_score(y_test, y_prob), 3))
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm).plot(cmap='Blues')
plt.title("Confusion Matrix")
plt.show()

feature_importance = pd.Series(model.feature_importances_[:10], index=[f"PC{i+1}" for i in range(10)]).sort_values(ascending=False)
feature_importance.plot(kind='bar', figsize=(6,4))
plt.title("Top PCA Feature Importances")
plt.ylabel("Importance")
plt.show()

kmeans = KMeans(n_clusters=5, random_state=42)
df['song_category'] = kmeans.fit_predict(X_pca)
print(df['song_category'].value_counts())

pca_df = pd.DataFrame(X_pca, columns=[f"PC{i+1}" for i in range(X_pca.shape[1])])
pca_df["category"] = df["song_category"]
sns.scatterplot(x="PC1", y="PC2", hue="category", data=pca_df, palette="tab10")
plt.title("PCA Song Clusters")
plt.show()

joblib.dump(model, "hit_song_model_v4.pkl")
joblib.dump(pca, "pca_v4.pkl")
joblib.dump(scaler, "scaler_v4.pkl")
joblib.dump(kmeans, "kmeans_v4.pkl")

new_song = pd.DataFrame([{
    'duration_ms': 205000, 'explicit': 0, 'danceability': 0.86, 'energy': 0.92,
    'key': 7, 'loudness': -4.3, 'mode': 1, 'speechiness': 0.05,
    'acousticness': 0.08, 'instrumentalness': 0.0002, 'liveness': 0.12,
    'valence': 0.81, 'tempo': 124, 'time_signature': 4,
    'duration_min': 205000/60000, 'energy_valence_ratio': 0.92/(0.81+0.01),
    'acousticness_inverse': 1-0.08, 'dance_energy_mix': (0.86+0.92)/2,
    'tempo_norm': (124 - df['tempo'].mean()) / df['tempo'].std()
}])

X_new = scaler.transform(new_song)
X_new_pca = pca.transform(X_new)
hit_prob = model.predict_proba(X_new_pca)[0, 1]
cluster_pred = kmeans.predict(X_new_pca)[0]

print("Hit Probability:", round(hit_prob, 3))
print("Hit Song ✅" if hit_prob >= 0.35 else "Not a Hit ❌")
print("Category:", cluster_pred)
```

```python
explicit_counts = df['explicit'].value_counts()
mode_counts = df['mode'].value_counts()
ts_counts = df['time_signature'].value_counts()
top_genres = df['track_genre'].value_counts().head(5)
top_artists = df['artists'].value_counts().head(5)

# Set up 3x2 grid for 5 pies
fig, axes = plt.subplots(3, 2, figsize=(14, 14))
axes = axes.flatten()

# 1️⃣ Explicit vs Non-Explicit
axes[0].pie(explicit_counts, labels=explicit_counts.index, autopct='%1.1f%%',
            colors=['#66b3ff','#ff9999'])
axes[0].set_title('Explicit vs Non-Explicit Songs')

# 2️⃣ Mode (Major / Minor)
axes[1].pie(mode_counts, labels=['Major (1)', 'Minor (0)'], autopct='%1.1f%%',
            colors=['#99ff99','#ffcc99'])
axes[1].set_title('Major vs Minor Key Songs')

# 3️⃣ Time Signature
axes[2].pie(ts_counts, labels=ts_counts.index, autopct='%1.1f%%',
            colors=sns.color_palette('muted'))
axes[2].set_title('Distribution of Time Signatures')

# 4️⃣ Top 5 Track Genres
axes[3].pie(top_genres, labels=top_genres.index, autopct='%1.1f%%', startangle=140,
            colors=sns.color_palette('pastel'))
axes[3].set_title('Top 5 Track Genres')

# 5️⃣ Top 5 Artists (by Number of Songs)
axes[4].pie(top_artists, labels=top_artists.index, autopct='%1.1f%%', startangle=140,
            colors=sns.color_palette('coolwarm'))
axes[4].set_title('Top 5 Artists by Number of Songs')

# Remove the 6th empty subplot
fig.delaxes(axes[5])

plt.tight_layout()
plt.show()
```

```python
# Recompute engineered features (if not saved in the CSV)
df['popularity_level'] = pd.cut(df['popularity'], bins=[0, 40, 70, 100], labels=['Low', 'Medium', 'High'])
df['duration_category'] = pd.cut(df['duration_min'], bins=[0, 2.5, 4.5, 15], labels=['Short', 'Medium', 'Long'], include_lowest=True)

df['mood_intensity'] = (df['valence'] * df['energy']) - (df['loudness'] / 60)
df['replay_value_score'] = (
    df['danceability']*0.35 + 
    df['energy']*0.35 + 
    (1 - df['instrumentalness'])*0.20 +
    (df['duration_min'].between(2.5, 4.5).astype(int) * 0.10)
)
df['party_index'] = df['energy']*0.4 + df['danceability']*0.4 + (1 - df['speechiness'])*0.2
df['sadness_score'] = (1 - df['valence']) * (1 - df['energy'])
df['production_complexity'] = df['instrumentalness']*0.5 + df['speechiness']*0.3 + df['liveness']*0.2
df['artist_popularity_deviation'] = df['popularity'] - df.groupby('artists')['popularity'].transform('mean')
df['vibiness_index'] = df['danceability']*0.4 + df['valence']*0.3 + (df['tempo'].between(90,140).astype(int) * 0.3)
df['skip_likelihood'] = ((df['duration_min'] > 5).astype(int)*0.3 + (1 - df['energy'])*0.3 + (1 - df['danceability'])*0.2 + df['instrumentalness']*0.2)
```

# HYPOTHESTSIS

```python
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
alpha = 0.05
```

```python
explicit = df[df['explicit'] == 1]['popularity']
non_explicit = df[df['explicit'] == 0]['popularity']
t_stat, p_val = stats.ttest_ind(explicit, non_explicit, equal_var=False)
print("T-Test: Explicit vs Non-Explicit Popularity")
print(f"t-statistic = {t_stat:.3f}, p-value = {p_val:.4f}")
print("→ Reject H₀" if p_val < alpha else "→ Fail to Reject H₀")
print("-" * 60)
```

```python
anova_result = stats.f_oneway(*[group['popularity'].values for _, group in df.groupby('track_genre') if len(group) > 2])
print("2️⃣ ANOVA: Popularity Across Genres")
print(f"F-statistic = {anova_result.statistic:.3f}, p-value = {anova_result.pvalue:.4f}")
print("→ Reject H₀" if anova_result.pvalue < alpha else "→ Fail to Reject H₀")
print("-" * 60)
```

```python
contingency = pd.crosstab(df['popularity_level'], df['explicit'])
chi2, p_chi, dof, expected = stats.chi2_contingency(contingency)
print("4️⃣ Chi-Square: Popularity Level vs Explicit")
print(f"Chi² = {chi2:.3f}, p-value = {p_chi:.4f}")
print("→ Reject H₀" if p_chi < alpha else "→ Fail to Reject H₀")
print("-" * 60)
```

```python
model = sm.OLS(df['popularity'], sm.add_constant(df[['replay_value_score', 'party_index']])).fit()
print("5️⃣ Multiple Linear Regression: Predicting Popularity")
print(model.summary())
print("-" * 60)
```

```python
from statsmodels.multivariate.manova import MANOVA

manova = MANOVA.from_formula('energy + valence + danceability ~ duration_category', data=df)
print("6️⃣ MANOVA: Song Features vs Duration Category")
print(manova.mv_test())
print("-" * 60)
```

```python
anova_time = stats.f_oneway(*[group['popularity'].values for _, group in df.groupby('time_signature') if len(group) > 2])
print("ANOVA: Popularity Across Time Signatures")
print(f"F = {anova_time.statistic:.3f}, p = {anova_time.pvalue:.4f}")
print("→ Fail to Reject H₀" if anova_time.pvalue >= 0.05 else "→ Reject H₀")
```

```python
corr2, p_val7 = stats.pearsonr(df['vibiness_index'], df['production_complexity'])
conclusion7 = "Reject H₀" if p_val7 < 0.05 else "Fail to Reject H₀"

print(f"Correlation Coefficient: {corr2:.4f}")
print(f"p-value: {p_val7:.4f}")
print(f"Conclusion: {conclusion7}")
```
